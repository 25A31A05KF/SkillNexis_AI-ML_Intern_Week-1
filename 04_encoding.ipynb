```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OneHotEncoder

data = {
    "Name": ["Rahul", "Ananya", "Vikram", "Priya", "Arjun", "Sneha"],
    "City": ["Hyderabad", "Bengaluru", "Chennai", "Hyderabad", "Mumbai", "Pune"]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)
```

```python
label_encoder = LabelEncoder()

df["City_Label"] = label_encoder.fit_transform(df["City"])

print("After Label Encoding:")
print(df)
```

```python
one_hot_encoder = OneHotEncoder(sparse_output=False)

city_encoded = one_hot_encoder.fit_transform(df[["City"]])

city_encoded_df = pd.DataFrame(
    city_encoded,
    columns=one_hot_encoder.get_feature_names_out(["City"])
)

print("After One-Hot Encoding:")
print(city_encoded_df)
```
