```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Rahul", "Ananya", "Vikram", "Priya", "Arjun", "Sneha", "Kiran", "Meera"],
    "Age": [20, 21, np.nan, 22, 20, np.nan, 23, 19],
    "Marks": [85, 91, 78, np.nan, 95, 82, 76, 89],
    "City": ["Hyderabad", "Bengaluru", "Chennai", "Hyderabad",
             "Mumbai", "Pune", "Delhi", "Chennai"]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)
```

```python
print("Missing values before handling:")
print(df.isnull().sum())
```

```python
# Fill missing Age values with the median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing Marks values with the mean
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

print("Data after handling missing values:")
print(df)
```

```python
print("Missing values after handling:")
print(df.isnull().sum())
```
